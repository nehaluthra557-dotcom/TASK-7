# Task 7: Basic Sales Summary from a Tiny SQLite Database

```python
import sqlite3
import pandas as pd
import matplotlib.pyplot as plt
```

```python
conn = sqlite3.connect("sales_data.db")
print("SQLite database connected successfully!")
```

```python
query = """
SELECT product,
       SUM(quantity) AS total_qty,
       SUM(quantity * price) AS revenue
FROM sales
GROUP BY product
ORDER BY revenue DESC;
"""
df = pd.read_sql_query(query, conn)
print(df.to_string(index=False))
```

```python
query2 = """
SELECT SUM(quantity) AS total_quantity_sold,
       SUM(quantity * price) AS total_revenue
FROM sales;
"""
totals = pd.read_sql_query(query2, conn)
print(totals.to_string(index=False))
```

```python
df.plot(kind="bar", x="product", y="revenue", legend=False, figsize=(9,5))
plt.title("Revenue by Product")
plt.xlabel("Product")
plt.ylabel("Revenue (INR)")
plt.xticks(rotation=30, ha="right")
plt.tight_layout()
plt.savefig("sales_chart.png", dpi=160)
plt.show()
```

```python
conn.close()
print("Database connection closed.")
```

## Key Observations

SQL aggregation was used to calculate total quantity and revenue by product. The results were loaded into Pandas and visualized using a basic Matplotlib bar chart.
